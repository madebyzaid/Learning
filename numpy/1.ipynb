ARRAY INDEXING

Same as lists. But note that numpy arrays have a fixed data type so if you assign a floating point value to an element in numpy array of integers, the value will be silently truncated.

```python
arr = np.array([[3, 5, 2, 4],
                [7, 6, 8, 8],
                [1, 6, 7, 7]])
```

```python
arr[1, 3] = 0
arr
```

ARRAY SLICING

General Syntax: arr[start: stop : step, start: stop: step, ...]
where commas separate dimensions

The default values are:

start = 0
stop = size of that dimension
step = 1

```python
np.arange(10)[0:10:2]
```

```python
np.array([[34, 65, 98],
          [36, 86, 29],
          [20, 37, 95]])[0:2, 1:2]
```

Accessing rows and columns

```python
x = np.array([[34, 65, 98],
              [36, 86, 29],
              [20, 37, 95]])
```

```python
x[:, 1] # second column of x
```

```python
x[1, :] # second row of x
```

```python
x[1] # second row of x
```

Subarrays as no-copy views

Array slices return views rather than copies of the array data. Different from python's list slicing.

Modifying slice modifies the slice in the original array itself

```python
g = np.array([[1, 5, 2, 4],
              [7, 6, 8, 8],
              [1, 6, 7, 7]])
```

```python
g[0:2, 2] # first two rows of third column
```

```python
g[0:2, 2] = [99, 100]
g
```
